**1.What are Embeddings?**

Embeddings are numerical representations of text,images,audio or any data.For text (usually our focus), an embedding is:

- A vector (a long list of numbers)

- Generated by a model

- Designed so that similar meanings --> similar numbers

**2.Why Do We Use Embeddings?**

Because computers dont understand text directly--they understand numbers.

Embeddings allow sysytems to:

 - Search semantically ("meaning_based search")

 - Build recommendation sysytems

 - Measure similarity between texts

 - Cluster documents

 - RAG(Retrieval-Augmented Generation), including ChatGroq workflows

**3.What Is a Similarity Score?**

Once you have embeddings,you measure similarity using a mathematical formula.

The most common one is:

**Cosine Similarity**

It measures the angle between two vectors.

    - 1.0 -> identical meaning

    - 0.0 -> unrelated

    - -1.0 -> opposite meaning

```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
# Open Source model NO api-key
#For testing purpose
import warnings
warnings.filterwarnings('ignore')
model=SentenceTransformer("all-MiniLM-L6-v2")
```

```python
#converting sentences to vectors
sentences=[
    "I like apples.",
    "I enjoy eating apples.",
    "I drive a motorcycle."
    
]
```

```python
embeddings=model.encode(sentences)
embeddings
```

```python
def cosine_sim(v1,v2):
    return np.dot(v1,v2)/(np.linalg.norm(v1)*np.linalg.norm(v2))
    
```

```python
cosine_sim(v1,v2)
```
